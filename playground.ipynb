```python
from bunq.sdk.context.api_context import ApiContext
from bunq.sdk.context.bunq_context import BunqContext
from bunq import ApiEnvironmentType
from bunq.sdk.model.generated.endpoint import MonetaryAccountBankApiObject, CardDebitApiObject, PaymentApiObject,BunqMeTabResultResponseApiObject,BunqMeTabApiObject, BunqMeTabEntryApiObject, BunqMeTabEntryApiObject, CardApiObject
from bunq.sdk.model.generated.object_ import AmountObject, PointerObject, NotificationFilterObject
from bunq import Pagination
import time

# Load API key
with open("api.txt", "r") as f:
    api_key = f.read().strip()

# Create an API context for production
api_context = ApiContext.create(
    ApiEnvironmentType.SANDBOX,
    api_key,
    "My Device Description"
)
```

```python
# Save the API context to a file for future use
api_context.save("bunq_api_context.conf")
# Load the API context into the SDK
BunqContext.load_api_context(api_context)
user_context = BunqContext.user_context()
# Get the user ID
user_id = user_context.user_id
print(user_id)
primary_account: MonetaryAccountBankApiObject = user_context.primary_monetary_account
print("monetary account", primary_account.id_)
```

```python
# List of budget categories
categories = ["groceries", "cafe", "shopping", "skincare", "fitness", "nightlife"]

# Dictionary to store category to account ID mapping
category_to_account_id = {}

# Loop to create accounts
for category in categories:
    print(f"Creating account: {category}")
    
    new_account_response = MonetaryAccountBankApiObject.create(
        currency="EUR",
        description=category.capitalize(),
        daily_limit=AmountObject("100.00", "EUR")
    )
    
    account_id = new_account_response.value
    print(f"✅ Created '{category}' → Monetary Account ID: {account_id}")
    category_to_account_id[category] = account_id

# Output the final mapping
print("\n📌 Final category-to-account ID mapping:")
for category, account_id in category_to_account_id.items():
    print(f"{category}: {account_id}")
```

```python
# STEP 2: Get user details and primary account alias
user_context = BunqContext.user_context()
user_person = user_context.user_person
user_id = user_context.user_id

# Use the primary account alias (IBAN) for the card
primary_account = user_context.primary_monetary_account
alias_pointer = primary_account.alias[0]  # IBAN Pointer

# STEP 3: Create the virtual debit card
response = CardDebitApiObject.create(
    second_line="GeoCard AI",
    name_on_card=user_person.display_name,
    alias=alias_pointer,
    type_="MASTERCARD",
    product_type="MASTERCARD_DEBIT",
    order_status="VIRTUAL_DELIVERY"
)

print("✅ Card created!")
print("Card ID:", response.value.id_)
```
